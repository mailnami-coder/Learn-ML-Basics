**Identifiers**
a user-defined name used to identify and reference program elements like variables, functions, classes, modules, or other objects.

**Rule1:**
Allowed Characters: Identifiers can only contain alphanumeric characters (a-z, A-Z, 0-9) and underscores (_). Since Python 3, they can also include non-ASCII Unicode letters (like café)

```python
#Variable
a=100
#Function
def add():
  pass
#Class
class student:
  pass
#Object
sushma=student()

```

**Rule2:**
Cannot Start with a Digit: An identifier can start with a letter or an underscore, but never a number (e.g., 1variable is invalid, but variable1 is valid)

```python
#Variable
#Incorrect
9a=100
#Correct
a9=100

#Function
#Incorrect
def 9add():
  pass
#correct
def add9():
  pass

#Class
#Incorrect
class 9student:
  pass
#correct
class student9:
  pass
```

**Rule3:**
No Special Characters or Spaces: Punctuation marks and symbols like @, $, %, !, or spaces are strictly prohibited

**Rule4:**
Reserved Keywords: You cannot use Python's built-in keywords (such as if, for, while, class, or def) as identifiers because they have a predefined meaning in the syntax.

**Rule5:**
Case-Sensitivity: Python treats uppercase and lowercase letters differently. For example, myVar, MyVar, and MYVAR are considered three completely separate identifiers.
