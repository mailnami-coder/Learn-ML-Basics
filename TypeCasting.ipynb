# Typecasting
It the process of converting one data type to another datatype

Syntax:
datatype(var/value)

```python
#Int Typecasting - It is the process of converting int data type into other data type.
#Int can be convertible into all SVDT data types
#When we try to convert into collection DT the conversion is possible only for string DT
#It is not possible for other collection data type

a=10
print(type(a))
print(type(float(a)))
print(type(str(a)))
print(type(complex(a)))
print(type(bool(a)))
print(type(str(a)))
#Below is not possible. Uncomment and run to see the error.
#print(type(list(a)))
#print(type(tuple(a)))
#print(type(set(a)))
#print(type(dict(a)))
```

```python
#Float type Casting - Process of converting float DT to other DT's
#Float can be converted to any SVDT
#In Collection DT float can converted only to string DT and other MVDT is not possible.

a=10.5
print(type(a),a)
print(type(int(a)),int(a))
print(type(str(a)),str(a))
print(type(complex(a)),complex(a))
print(type(bool(a)),bool(a))
print(type(str(a)),str(a))

#Below is not possible. Uncomment and run to see the error.
#print(type(list(a)))
#print(type(tuple(a)))
#print(type(set(a)))
#print(type(dict(a)))
```

```python
#Complex DT casting
#Complex DT is convertible only to the bool and string DT, rest of the DT is not possible.

a=10+4j
print(bool(a),type(bool(a)))
print(str(a),type(str(a)))

#All below statements will cause error
#print(int(a),type(int(a)))
#print(float(a),type(float(a)))
#print(list(a),type(list(a)))
```

```python
#Boolean Typecasting - It can be convertible into all the SVDT
#Incase of MVDT only string is possible, other MVDT is not possible.
```

```python
a=True
print(type(a))
print(type(str(a)),str(a))
b=str(a)
print(type(b))
if b=="True":
    print("True")
else:
    print("False")
#
```

# String Typecasting

String can be converted into all the SVDT, but it should only have base of 10 values

String can be converted to all the collection datatype except Dictionary DT

```python
a='gowri'
b='9'
print(list(a),type(a))
print(tuple(a),type(a))
print(set(a),type(a))
print(int(b),type(int(b)))
print(float(b),type(float(b)))
print(complex(b),type(complex(b)))
print(bool(a),type(bool(a)))
#print(dict(a),type(a))
```

# List TypeCasting
List is convertible only to bool DT,other SVDT is not possible

List can be converted to all MVDT but for Dictionary data type it should have nested collection with two values.

```python
a=[1,2,3,4]
print(bool(a),type(bool(a)))
print(tuple(a),type(tuple(a)))
print(set(a),type(set(a)))
print(str(a),type(str(a)))
b=[[1,11],[2,22],[3,33]]
print(dict(b),type(dict(b)))
```

##Tuple Typecasting
Tuple can be converted only to bool DT, other SVDT is not possible

Tuple can be converted to all MVDT, incase of DICT it should be nested collection with two values.

```python
a=(1,2,4,56,6)
print(bool(a),type(bool(a)))
print(str(a),type(str(a)))
print(list(a),type(list(a)))
print(set(a),type(set(a)))
b=((1,10),(2,20),(3,30))
print(dict(b),type(dict(b)))
b=((4,40),(5,50))
print(dict(b),type(dict(b)))
```
